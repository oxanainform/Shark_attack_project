```python
import pandas as pd

# Load the data and keep records from 2000 onward.
df = pd.read_csv("/Users/foxy/Downloads/GSAF5.csv", low_memory=False)
df["Year"] = pd.to_numeric(df["Year"], errors="coerce")
df_2000_onward = df[df["Year"] >= 2000].copy()

# Normalize full month names and common abbreviations to full month names.
month_aliases = {
    "jan": "January", "january": "January",
    "feb": "February", "february": "February",
    "mar": "March", "march": "March",
    "apr": "April", "april": "April",
    "may": "May",
    "jun": "June", "june": "June",
    "jul": "July", "july": "July",
    "aug": "August", "august": "August",
    "sep": "September", "sept": "September", "september": "September",
    "oct": "October", "october": "October",
    "nov": "November", "november": "November",
    "dec": "December", "december": "December",
}
months = [
    "January", "February", "March", "April", "May", "June",
    "July", "August", "September", "October", "November", "December",
]
month_pattern = r"\b(" + "|".join(month_aliases) + r")\b"
extracted_months = (
    df_2000_onward["Date"]
    .astype("string")
    .str.lower()
    .str.extract(month_pattern, expand=False)
)
df_2000_onward["Month"] = pd.Categorical(
    extracted_months.map(month_aliases),
    categories=months,
    ordered=True,
)

# Inspect unrecognized date formats before removing the original Date column.
print(df_2000_onward.loc[df_2000_onward["Month"].isna(), "Date"])
df_2000_onward = df_2000_onward.drop(columns=["Date"])
```
