```python
import pandas as pd
import numpy as np
```

```python
shark_attacks = pd.read_csv("GSAF5.csv")
display(shark_attacks)
```

```python
print("\nColumns:")
display(shark_attacks.columns.tolist())
print("\nInfo:")
display(shark_attacks.info())
print("\nData types:")
display(shark_attacks.dtypes)
```

```python
attacks_reduced = shark_attacks.drop(columns=["State","Name",'Time',"Source",'pdf', 'href formula', 'href', 'Case Number', 'Case Number.1',
       'original order', 'Unnamed: 21', 'Unnamed: 22'])

display(attacks_reduced)
```

```python
print("\nMissing values:")
display(attacks_reduced.isna().sum())
```

```python
without_missing_sex = attacks_reduced.dropna(subset=["Sex"])
display(without_missing_sex)
```

```python
print("\nMissing values:")
display(without_missing_sex.isna().sum())
```

```python
without_missing_sex.columns = (
    without_missing_sex.columns #takes the column names of the DataFrame
    .str.strip() #removes leading and trailing whitespace from the column names
    .str.lower() #converts the column names to lowercase
    .str.replace(" ", "_") #replaces spaces with underscores in the column names
)

print(without_missing_sex.columns.tolist())
```

```python
for column in without_missing_sex.describe(include=["O"]).columns:
    print(without_missing_sex[column].value_counts())
    print()
```

```python
#filtering dataframe, only data since year 2000
without_missing_sex["year"] = pd.to_numeric(without_missing_sex["year"], errors="coerce")

attacks_reduced_2000_onward = without_missing_sex[without_missing_sex["year"] >= 2000].copy()
display(attacks_reduced_2000_onward)
```

```python
print("\nMissing values:")
display(attacks_reduced_2000_onward.isna().sum())
```

```python
print("\nDuplicated rows:")
print(attacks_reduced_2000_onward.duplicated().sum())
```

```python
print("\nDate:")
display(attacks_reduced_2000_onward["date"].value_counts())
```

```python

# Normalize full month names and common abbreviations to full month names.
month_aliases = {
   "jan": "January", "january": "January",
    "feb": "February", "february": "February",
    "mar": "March", "march": "March",
    "apr": "April", "april": "April",
    "may": "May",
    "jun": "June", "june": "June",
    "jul": "July", "july": "July",
    "aug": "August", "august": "August",
    "sep": "September", "sept": "September", "september": "September",
    "oct": "October", "october": "October",
    "nov": "November", "november": "November",
    "dec": "December", "december": "December",
}
months = [
    "January", "February", "March", "April", "May", "June",
    "July", "August", "September", "October", "November", "December",
]
month_pattern = r"\\b(" + "|".join(month_aliases) + r")\\b"
extracted_months = (
    attacks_reduced_2000_onward["date"]
    .astype("string")
    .str.lower()
    .str.extract(month_pattern, expand=False)
)
attacks_reduced_2000_onward["month"] = pd.Categorical(
    extracted_months.map(month_aliases),
    categories=months,
    ordered=True,
)

# Inspect any dates for which no month name or abbreviation was found.
print(attacks_reduced_2000_onward.loc[attacks_reduced_2000_onward["month"].isna(), "date"].head())
attacks_reduced_2000_onward = attacks_reduced_2000_onward.drop(columns=["date"])
display(attacks_reduced_2000_onward)

#display unique values months
display(attacks_reduced_2000_onward["month"].unique())
```
